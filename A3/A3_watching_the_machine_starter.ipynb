## 4. Synthesis & reflection

*Bring the probe and the observation together into one account. What does this mean for trustworthy, reproducible research — and what will you change about how you use and verify AI in your own work?*

_Write here._

My probe and my observation point the same direction, on a short, mechanical, easy-to-check task, the LLMs were right every time. All four runs across ChatGPT and Gemini returned 9, matching my ground truth, and both participants accepted the same answer from an LLM without further checks. Their reason was that counting words is “manual-ish” work where it is hard to go wrong. That reasoning held up here, but only because the task happened to be simple. Four runs on one 3,763-character passage show that this task is reliable, not that LLMs are. Trust that is justified on this task could easily become over-trust on a task where the answer isn’t obvious, and neither my participants nor I would necessarily notice the difference.

The more important finding is how we all decide when to check. In Section 1 I wrote that I scrutinize output more when the stakes are high, and both participants said the same thing: low stakes means little checking, and high stakes means checking by hand or with other tools. That is a sensible rule, but it depends on correctly judging the stakes, and on being able to verify at all. My check took minutes because I wrote the regex and read all nine hits. In my coding work, where I’m less familiar with the code, I can’t verify in the same way, and I have assumed correctness because the output comes from “a machine talking to another machine.” My probe doesn’t test that assumption, so I shouldn’t treat it as confirmed.

The log also shows that I can’t see how the models reached their answer. My notes record that they were “thinking” and “running code,” so I don’t know whether the correct counts came from reasoning or from executed code. That limits what I can claim about reproducibility. There is also an ethical side: a confident, correct answer discourages checking, and the people least able to verify (those less familiar with the method) are the ones most likely to rely on it. My Section 1 observation that models tend to agree with whatever I say is a related reliability problem that wrong answers don’t capture.

What I will change: (1) For code I don’t fully understand, I will write my own small check, as I did here, before relying on it. (2) I will record the model, version, and time for anything I rely on, so I can reproduce it. (3) I will treat “stakes” as something to judge before I ask, not after I see a plausible answer.
